```python
def calc_fee( card_type  , charge_amount  ):
    if card_type == 'Visa':
        fee = 0.018*charge_amount
    elif card_type == 'Mastercard':
        fee = 0.0195*charge_amount
    elif card_type == 'Amex':
        fee = 0.0205*charge_amount
    return(  fee +charge_amount +0.25  )
```

### Useful numpy functions

```python
# 1) draws samples from a Uniform distribution U(a,b). Default: U(0,1).
# np.random.uniform(a, b [,size])
# [,size] means that size is optional. it is 1 by default.
np.random.uniform(1,10,5)
```

```python
# 2) draws samples from a Normal distribution N(mu,sigma)
# sigma is standard deviation, not variance!
# np.random.normal(mu, sigma [,size])
np.random.normal(5,1)
```

```python
# 3) draws samples from an Exponential distribution Exp(lambda)
# np.random.exponential(scale[, size])
# note that scale = 1/lambda = E(X)! scale is not equal to lambda
np.random.exponential(5, 100)
```

```python
# 4) draws samples from a Gamma distribution Gamma(k, theta)
# np.random.gamma(shape, scale, size)
# shape = k, scale = theta. 
# E(X) = k·theta, var(X) = k·theta^2
np.random.gamma(2, 3, 10)
```

```python
# a) Uniform(10, 40)
two_a = np.random.uniform(10,40,100000)
two_a_table = Table().with_columns("Uniform", two_a)
two_a_table.hist(0,bins = np.arange(10,40,0.4))
```

```python
# b) Normal(25,5)
two_b = np.random.normal(25,5,100000)
two_b_table = Table().with_columns("Normal", two_b)
two_b_table.hist(0,bins = np.arange(10,40,0.4))
```

```python
# c) Exponential(25)
two_c = np.random.exponential(25,100000)
two_c_table = Table().with_columns("Exponential", two_c)
two_c_table.hist(0,bins = np.arange(0,80,2))
```

```python
# d) Gamma(10,25)
two_d = np.random.gamma(10,25,100000)
two_d_table = Table().with_columns("Gamma", two_d)
two_d_table.hist(0,bins = np.arange(5,60,1))
```

### Problem 3

```python
# a) Uniform(10, 40)
table = np.zeros(100000)
for n in range(0,100000):
    table[n] = np.mean(np.random.uniform(10, 40, 100))
three_a_table = Table().with_columns('Uniform', table)
three_a_table.hist(0,bins = np.arange(20,30,0.1))
```

```python
# c) Exponential(25)
table = np.zeros(100000)
for n in range(0,100000):
    table[n] = np.mean(np.random.exponential(25, 100))
three_c_table = Table().with_columns('Exponential', table)
three_c_table.hist(0,bins = np.arange(15,35,0.5))
```
