### Ranges

```python
# numpy.arange(): generate some specific arrays

# NOT 'arrange'! IT IS 'arange'!


# numpy.arange(n): array([0,1,2,...,n-1])
numpy.arange(5)

# Usually we use, e.g., for i in numpy.arange(10)
```

```python
# numpy.arange(a,b): array([a,a+1,...,b-1])
numpy.arange(3,11)
```

```python
# numpy.arange(a,b,s): array([a,a+s,a+2s,..., UP TO b-1])
# can be b-2, b-3,...

numpy.arange(1,11,2)
```
